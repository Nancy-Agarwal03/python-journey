```python
#grouping and aggrigation
df.groupby("country")["income"].mean()
df.groupby("country")["income"].min()
df.groupby("country")["income"].max()

df.groupby("gender")["income"].mean()
df.groupby("gender")["income"].min()
df.groupby("gender")["income"].max()

df.groupby("country")["income"].agg(["min","mean","max"])

df.groupby("country")["income"].agg(min_salary="min",avg_salary="mean",max_salary="max")
df.groupby("country").agg({
    "income":"mean",
    "age":"mean"
})
df.groupby("country").agg(
    max_salary=("income","max"),
    mean_age=("age","mean")
)
```

```python
# MELT & PIVOT
df = pd.DataFrame({
    "country":["USA","USA","INDIA","INDIA"],
    "year":[2021,2023,2021,2023],
    "sales":[100,200,110,170],
    "profit":[20,50,30,25],
})


melted_df=df.melt(
    id_vars=["country","year"],
    value_vars=["sales","profit"],
    var_name="metrics",
    value_name="value"
)

melted_df.pivot(
    index=["country","year"],
    columns="metrics",
    values="value"
)
```

```python
#Merge and join
df_customers= pd.DataFrame({ 
    "customer_id": [1,2,3,4],
    "name":["Adam","Bob","Charlie","Dave"]
}) 

df_orders= pd.DataFrame({ 
    "order_id": [101,102,103,104],
    "customer_id":[2,1,4,5],
    "amount":[250,120,300,180]
})

pd.merge(df_customers,df_orders,on="customer_id")# Inner Join
pd.merge(df_customers,df_orders,on="customer_id", how="left")#Left Join
pd.merge(df_customers,df_orders,on="customer_id", how="right")# Right Join
pd.merge(df_customers,df_orders,on="customer_id", how="outer")# OuterJoin
```
